```python
df = pd.read_csv("../data/transactions.csv")
df
```

```python
df.sort_values("amount", ascending=False)
```

```python
df["amount"].describe()
```

```python
q1 = df["amount"].quantile(0.25)
q3 = df["amount"].quantile(0.75)

iqr = q3 - q1

q1, q3, iqr
```

```python
upper_threshold = q3 + 1.5 * iqr
upper_threshold
```

```python
df[df["amount"] > upper_threshold]
```

```python
## Isolation Forest
```

```python
from sklearn.ensemble import IsolationForest

model = IsolationForest(
    contamination=0.06,
    random_state=42,
)

df["ml_prediction"] = model.fit_predict(df[["amount"]])

df[["transaction_id", "amount", "ml_prediction"]]
```
